Logging with Multiple Loggers

```python
import logging

## Create logger  for module 1

logger1 = logging.getLogger('module1')

logger1.setLevel(logging.DEBUG)

# create logger for module 2

logger2 = logging.getLogger("module2")

logger2.setLevel(logging.WARNING)


# create logging settings
logging.basicConfig(
    level = logging.DEBUG,
    format = '%(asctime)s - %(name)s - %(levelname)s - %(message)s',
    datefmt = '%Y-%m-%d %H:%M:%S',
)
```

```python
# log message with different loggers

logger1.debug("This is a debug message for module1")
logger2.warning("This is a warning message for module2")
logger2.error("This is a error message for module2")
logger2.critical("This is a critical message for module2")
logger1.info("This is an info message for module1")
```
