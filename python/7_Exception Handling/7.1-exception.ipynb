Exceptions

- Exception Handling in python allows you to handle errors gracefully and take corrective actions without stopping the execution of the program.

what are exceptions?
- Exceptions are events that disrupt the normal flow of a program.
- They occure when an error is encountered during program execcution.

Common Exceptions:
1. ZeroDivisonError: Dividing by zero
2. FileNotFoundError: file not found
3. ValueError: invalid value
4. TypeError: invalid type

```python
a = 10 # this work 
a = b # but this give us error/exception bcoz 'b' is not defined --> NameError 
```

```python
# Exception try, except block

try:
    a=b
except:
    print("The variable has not been assigned")
```

```python
try:
    a=b
except NameError as ex:
    print(ex)
```

```python
result = 1/0 # this is zero division error
```

```python
# to handle it use try, except with zeroDivisionError

try:
    result = 10/0
except ZeroDivisionError as ex:
    print(ex)
    print("Please enter the denominator greater than 0")
```

All this exceptions like ZeroDivisionError, FileNotFountError, ValueError, TypeError are derived from main exception class

```python
try:
    result = 10/2
    a=b
except ZeroDivisionError as ex:
    print(ex)
    print("Please enter the denominator greater than 0")
except Exception as ex1: # this handle all the exceptions
    print(ex1)
    print("Main exception got caught here")

# here now we getting NameError not ZeroDivisionError, so ZeroDivisionError will not hande it
# so we have main class i.e., Exception
```

```python
try:
    num = int(input("Enter a number"))
    result = 10/num
except ValueError:
    print("This is not a valid number")
except ZeroDivisionError:
    print("Enter denominator greater then 0")
except Exception as ex:
    print(ex)
```

```python
## try, except, else block

try:
    num = int(input("Enter a number"))
    result = 10/num
except ValueError:
    print("This is not a valid number")
except ZeroDivisionError:
    print("You can't divide by zero")
except Exception as ex:
    print(ex)

else:
    print(f"the result is {result}")
```

```python
## try, except, else, finally block


try:
    num = int(input("Enter a number"))
    result = 10/num
except ValueError:
    print("this is not a valid number")
except ZeroDivisionError:
    print("You can't divide by zero")
except Exception as ex:
    print(ex)

else:
    print(f"the result is {result}")

finally:
    print("This is finally block, always execute, whether exception occurs or not")
```
