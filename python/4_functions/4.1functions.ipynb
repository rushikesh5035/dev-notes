### Functions

- A function is a block of code that performs a specific task. 
- functions help in organizing code, reusing code, and improving readability

```python
## Syntax

def function_name (parameters):
    """Decstring"""
    # Function body
    return "expression"
```

```python
## why functions?

# if you want to check num is even or odd - you would have to write the same logic multiple times
# functions allow you to write the logic once and reuse it multiple times

num = 24

if num % 2 == 0:
    print("Even")
else:
    print("Odd")

# this code is repeated multiple times, so we can create a function to check if a number is even or odd
```

```python
## Creating function to check number is even or odd

def even_odd(num):
    """This function finds even or odd"""
    if num % 2 == 0:
        print("Even")
    else:
        print("Odd")
```

```python
## Call function
even_odd(14) # output - even
even_odd(15) # output - odd
```

```python
## function with multiple parameter

def add(a, b):
    """This function adds two numbers"""
    return a + b

add(5, 5)
```

### Default paramters

```python
def greet(name):
    print(f"Hello, {name}! Welcome to the world of functions!")

greet("Alice")  # Output: Hello, Alice! Welcome to the world of functions!
```

```python
def greet(name="Guest"): # default parameter
    print(f"Hello, {name}! Welcome to the world of functions!")

greet() # Output: Hello, Guest! Welcome to the world of functions!
greet("Bob") # Output: Hello, Bob! Welcome to the world of functions!
```

#### Variable Length Arguments

- Positional and keywords Arguments

```python
def print_numbers(*args): # variable length arguments -> if N numbers of arguments
    for num in args:
        print(num)

print_numbers(1, 2, 3, 4, 5, "Rushi", 3.23) # Output: 1 2 3 4 5, "Rushi", 3.23
```

- Keywords Arguments

```python
## Keyword arguments (**kwargs) allow you to pass a variable number of keyword arguments to a function. 
## keyword arguments - parameters are in key-value pair format, and you can pass them in any order.

def print_details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_details(name="Alice", age=30, city="New York") # Output: name: Alice, age: 30, city: New York
```

```python
def print_info(*args, **kwargs):
    for val in args:
        print(f"Positional arguments: {val}")

    for key, value in kwargs.items():
        print(f"{key}: {value}")    


print_info(1, 2, "Rushi", 3.23, name="Alice", age=30, city="New York")
```

#### Return Statements

```python
def multiply(a, b):
    """This function multiplies two numbers"""
    return a * b

multiply(5, 5) # Output: 25
```

```python
## return mutilple values from function

def multiply(a, b):
    return a * b, a + b, a - b

multiply(5, 5) # Output: (25, 10, 0)
```
