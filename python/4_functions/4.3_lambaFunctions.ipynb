#### Lambda Function in Python

- Lambda function are small anonymous functions defined using **Lambda** keyword.
- They can have any number of arguments but only one expression. 
- They are commonly used for short operations or as arguments to higher-order functions.

#### Syntax

`lambda arguments: expression`

```python
def addition(a, b):
    return a + b
```

```python
addition(2, 3)
```

```python
# let's create a lambda function that does the same thing as the addition function above

addition = lambda a,b: a + b

type(addition) # # <class 'function'>
```

```python
addition(2, 3) # 5
```

```python
def even(num):
    if num % 2 == 0:
        return True
    else:
        return False

even(23)
```

```python
even = lambda num: num % 2 == 0

print(even(22)) # True
print(even(23)) # False
```

```python
def addition(a, b, c):
    return a + b + c

print(addition(2, 3, 4)) # 9
```

```python
addition = lambda a, b, c: a + b + c
addition(2, 3, 4)
```

map() -> applies a function to all items in a list

```python
numbers = [1, 2, 3, 4, 5]

squared_num = map(lambda x: x ** 2, numbers)

list(squared_num) # [1, 4, 9, 16, 25]
```
