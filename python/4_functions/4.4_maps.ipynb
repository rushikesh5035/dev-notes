map() function in python

- the map() function applies a given function to all items in an input list (or any other iterable) and return a map object (an iterator). 
- this is particularly useful for transforming data in a list comprehensively

```python
def square(x):
    return x*x

square(5)
```

```python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]

# suppose now want to iterate all the element in list and sqaure it

result = map(square, numbers)

list(result)
```

```python
## lambda function with map

list(map(lambda x:x*x, numbers))
```

map multiple iterables

```python
number1 = [1, 2, 3, 4, 5]
number2 = [6, 7, 8, 9, 10]

list(map(lambda x, y: x+y, number1, number2))
```

map to convert a list of strings to integers

```python
str_num = ['1', '2', '3', '4', '5']

int_num = list(map(int, str_num))

print(int_num) # [1, 2, 3, 4, 5]
```

```python
words=['hello', 'world', 'python', 'programming']

# Convert all strings to uppercase
uppercaase_words = list(map(str.upper, words))
print(uppercaase_words)  # ['HELLO', 'WORLD', 'PYTHON', 'PROGRAMMING']
```

```python
# this is a list of dictionaries
def get_name(person):
    return person['name']


people = [
    {'name': 'Alice', 'age': 30},
    {'name': 'Bob', 'age': 25},
    {'name': 'Charlie', 'age': 35}
]


list(map(get_name, people))  # ['Alice', 'Bob', 'Charlie']
```
