File Operation - Read And Write Files

```python
## Read a whole file

with open('example.txt', 'r') as file:
    content = file.read()
    print(content)
```

```python
## Read a file line by line

with open('example.txt', 'r') as file:
    for line in file:
        print(line.strip()) # strip() removes the newline character at the end of each line
```

```python
## Writing a file(overwriting existing content)

with open("example.txt", "w") as file:
    file.write("This is a new line of text.\n")
    file.write("This is another line of text.\n")
    
```

```python
## writing a file(append mode)
# a - append mode - not overwrite existing content, but add new content at the end of the file

with open("example.txt", "a") as file:
    file.write("Append operation taking place! \n")
```

```python
## Writing a list of lines to a file

lines = ["First line \n", "second line \n", "Third line \n"]
with open("example.txt", 'a') as file:
    file.writelines(lines) # writelines() writes a list of lines to the file
```

Binary files

```python
## Binary files

## Writing to a binay file --> wb (write binary)

data = b'\x00\x01\x02\x03\x04\x05\x06\x07\x08\x09'

with open('example.bin', 'wb') as file:
    file.write(data)
```

```python
## reading a binary file --> rb (read binary)

with open("example.bin", "rb") as file:
    content = file.read()
    print(content)
```

Practical Example

```python
## Read the content from source text file and write to a destination text file

#copying a text file
with open("example.txt", 'r') as source_file:
    content = source_file.read()

with open("destination.txt", 'w') as destination_file:
    destination_file.write(content)
```

```python
## Read a text file and count the number of lines, words, and characters

def count_text_file(file_path):
    with open(file_path, 'r') as file:
        lines = file.readline()
        line_count = len(lines)
        word_count = sum(len(line.split()) for line in lines)
        char_count = sum(len(line) for line in lines)
    return line_count, word_count, char_count

file_path = "example.txt"
lines, words, characters = count_text_file(file_path)
print(f"Lines: {lines}, Words: {words}, Characters: {characters}")
```

```python
## Writing and then reading a file
# w+ mode -> write and read mode, it will create the file if it does not exist, and overwrite the existing content if it does exist.

with open("example.txt", "w+") as file:
    file.write("This is a new line of text.\n")
    file.write("This is another line of text.\n")

    # move the file pointer to the beginning of the file before reading
    file.seek(0)

    # Read the content of the file
    content = file.read()
    print(content)
```
