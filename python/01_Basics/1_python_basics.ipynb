#### Syntax and Semantics in python

```python
## Basic syntax rule in python
## Case sensitivity - Python is case sensitive, which means that uppercase and lowercase letters are treated as distinct. For example, the variable name 'Variable' is different from 'variable'.


name = "Rushi"
Name = "Tele"

print(name)  # Output: Rushi
print(Name)  # Output: Tele

# both are the different variables and can hold different values.
```

```python
## Indentation 
## python uses indentation to define blocks of code. Consistent use of spaces (commonly 4 spaces) or a tab is required to indicate the level of nesting. Improper indentation will result in an IndentationError.

age = 32
if age >= 18:
    print("You are an adult.") # 4 spaces indentation or 1 tab is used here

print("This line is not indented.") # This line is not indented, so it is outside the if block.
```

```python
## This is a single line comment in python.

print("Hello, World!")  # This is an inline comment.
```

```python
## Line Continuation - In Python, you can use a backslash (\) to indicate that a statement continues on the next line. This is useful for breaking long lines of code into multiple lines for better readability.

total = 1 + 2 + 3 + \
        4 + 5 + 6 + \
        7 + 8 + 9
print(total)  # Output: 45
```

```python
## Multiple Statement on a single line

x = 5; y = 10; z = 15

print(y)
```

```python
## understand Semantics in python

## variable assignment

age=32 ## age is an integer variable
name = "Rushi" ## name is a string variable
```

```python
## Type inference
## Python is a dynamically typed language, which means that you don't need to explicitly declare the type of a variable. The interpreter infers the type based on the value assigned to the variable.

variable = 10
print(type(variable))  # Output: <class 'int'>

variable = "Hello"
print(type(variable))  # Output: <class 'str'>

## it means during the runtime, the type of the variable can change based on the value assigned to it.
```
