 ### Variables in Python

```python
## Declaring and assigning variables

age = 22
height = 5.5
name = "Rushikesh Tele"
is_student = False

# Printing the variables

print("Age:", age)
print("Height:", height)
print("Name:", name)
print("Is Student:", is_student)
```

```python
## Naming conventions in Python
# variable name should be descriptive
# they much start with a letter or an "_" (underscore) and contains letter, numbers and underscores
# variables names are case sensitive

# valid variables names
first_name = "John"
last_name = "Doe"

subject1 = "Math"
subject2 = "Science"

# invalid variables names

2age = 20
first-name = "rushi"
@name="Rushi"
```

```python
## case sensitivity
name = "Rushi"
Name = "Rushikesh"

# both are different variables due to case sensitivity

name == Name # false
```

### Variables types

```python
## python is dynamically typed language, means type of a variable is determined at runtime

age = 25 # int
height = 5.3 # float
name = "Rushi" # string
is_student = True # bool 

print(type(name))
```

### Type Checking and Conversion

```python
type(height)
```

```python
age = 25
print(type(age)) # int

## Type conversion
age_str = str(age) 
print(age_str) 
print(type(age_str)) # str
```

```python
age = '25'

print(int(age))
print(type(age))
print(type(int(age)))
```

```python
name = "Rushi"
int(name) # not able to convert because name is completely string
```

```python
height = 5.5

print(type(height))

print(int(height)) # converting float to int

print(float(int(height))) # int to float
```

```python
## Dynamic typing
# python allows the type of a variable to change as the program executes

var = 10
print(var, type(var))

var = "hello"
print(var, type(var))

var = 1045.34
print(var, type(var))
```

```python
## Input

age= input("what is the age?")
print(age, type(age)) # type will be str

marks = int(input("Enter marks"))
print(marks, type(marks))
```
