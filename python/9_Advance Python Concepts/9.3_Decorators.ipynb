Decorators

Decorators are a powerful and flexible feature in Python that allows you to modify the behavior of a function or class method.
they are commonly used to add functiona;ity to functions or methods without modifying their actual code.

But before going to Decorators let's first understand what is `function copy` and `closures`

#### Function Copy

```python
def Welcome():
    return "Welcome to the advanced python concepts"

Welcome()
```

```python
wel = Welcome # this creates a copy of the function object and assigns it to the variable wel
print(wel())  # we can call the function using the variable wel

del Welcome # we can also delete the above function using del keyword
print(wel()) # this will still work because wel is a copy of the function object and it is not deleted
```

#### Closures

```python
## closures - closures meaning function inside the functions

def main_welcome(message):
    msg = "Welcome to main welcome method"
    def sub_welcome():
        print("Welcome to sub welcome method")

        print(message) 

        print("please learn these concepts properly")

        print(msg) # we can also use the main method's properties inside sub method
    
    return sub_welcome()


# call method
main_welcome("Hey")
```

```python
def main_welcome_method(func):
    def sub_welcome_method():
        print("Welcome to sub welcome method")

        func("Welcome everyone to this topic") # we can also call the main method's properties inside sub method

        print("please learn these concepts properly")
    return sub_welcome_method()
```

```python
main_welcome_method(print)
```

#### **Decorators**

```python
## Create decorator manually

def main_welcome_method(func):
    
    def sub_welcome_method():
        print("Welcome to sub welcome method")
        func()
        print("please learn these concepts properly")
    return sub_welcome_method()
```

```python
def course_intro():
    print("This is the course introduction method")

course_intro()
```

```python
main_welcome_method(course_intro)
```

```python
@main_welcome_method # this is a decorator, created using `@` symbol, which is a shorthand for calling the main_welcome_method with course_intro as an argument

def course_intro():
    print("This is the course introduction method")
```

```python
## Another Examples with decorators

def my_decorator(func):
    def wrapper():
        print("Something is happening before the function is called.")
        func()
        print("Something is happening after the function is called.")
    return wrapper
```

```python
@my_decorator
def say_hello():
    print("Hello!")


# call method
say_hello()

# as soon as we call the say_hello method, it will first call the wrapper method, which will print the message before and after the function is called.
```

```python
## Decorators with arguments

def repeat(decorator_arg):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(decorator_arg):
                func(*args, **kwargs)
        return wrapper
    return decorator

```

```python
@repeat(5)
def say_hello():
    print("Hello!")

# call method
say_hello()
```
