Loops

1] for loop

```python
range(5) # it gives us number from 0 to 4, 5 is not included
```

```python
for i in range(5):
    print(i)
```

```python
for i in range(1, 6):
    print(i)
```

```python
for i in range(1, 10, 2): # it gives us number from 1 to 9, 10 is not included, and the step is 1/2 (jumping 2 numbers)
    print(i)
```

```python
for i in range(10, 1, -2): # it gives us number from 10 to 2, 1 is not included, and the step is -1
    print(i)
```

```python
## strings -> collection of characters

str = "Rushi Tele"

for i in str:
    print(i)
```

2] while loop

```python
## while loop continues to execute a block of code as long as a specified condition is true.

count = 4

while count <= 5:
    print(count)
    count += 1
```

```python
count = 0

while count % 2 == 0:
    print(count)
    count = count+1
```

Loop Control Statements

1) Break ->
break statement exits the loop prematurely

```python
for i in range(10):
    if i == 5:
        break
    print(i)
```

2) Continue -> skips the current iteration and continues with the nextx

```python
for i in range(10):
    if i % 2 == 0:
        continue # all the even num skips
    print(i) # prints all the odd numbers
```

3) Pass -> it is a null operatuo; it does nothing

```python
for i in range(5):
    if i == 3:
        print("Found 3!")
        pass # it does nothing, just a placeholder
    print(i)
```

Nested loops -> loop inside a loop

```python
for i in range(2):
    for j in range(3):
        print(f"i: {i} and j: {j}")
```

```python
# Examples - calculate the sum of first N natural num using while loop and for loop

# while loop

n = 10
sum = 0
count = 1

while count <= n:
    sum = sum + count
    count = count + 1

print("Sum of first 10 natural num", sum)
```

```python
## for loop

n = 10
sum = 0
count = 1

for i in range(11):
    sum = sum+i

print("Sum of first 10 natural num", sum)
```

```python
for num in range(1, 51):
    if num > 1:
        for i in range(2, num):
            if num % i == 0:
                break
        else:
            print(num)
```
