Importing Modules in Python: Modules and packages

- in python, modules and packages help organize and reuse code.

```python
import math

math.sqrt(16)  # 4.0
```

```python
from math import sqrt, pi

print(sqrt(25))  # 5.0
print(sqrt(36))  # 6.0 
print(pi)
```

```python
import numpy as np 
# np is alias for numpy

print(np.__file__)
```

```python
np.array([1, 2, 3])  # Creates a numpy array
```

```python
from math import * # import all the function inside math

print(sqrt(36))
print(pi)
```

importing package from another file

```python
from package.maths import addition, subtraction, multiplication, division

print(addition(2, 3))  # 5
print(subtraction(6, 2))  # 4
print(multiplication(6, 2))  # 12
print(division(6, 2))  # 3
```
