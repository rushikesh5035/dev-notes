#### Handling missing Values

```python
df.isnull().any()
```

```python
df.isnull().sum() # checks for missing values in each column of the DataFrame and returns the count of missing values for each column.
```

```python
df_filled = df.fillna(0) # fills missing values in the DataFrame with 0 and modifies the DataFrame in place.

df_filled
```

```python
## filling missing values with mean of the column

df['ADDRESSLINE2_fillNA'] = df['ADDRESSLINE2'].fillna(df['ADDRESSLINE2'].mean())
df
```
