##### Sets
- Sets - Sets are a build n data type in python used to store collections of unique items
- sets are the mutable
- they are unirdered, meaning that the elements do not follow a specific order, and they do not allow duplicate elements.
- sets are usefull for memnership tests, eliminating duplicate entries, and performing mathematical operations like union, intersection, and difference.

Creating a Set

```python
my_set = {1, 2, 3, 4, 5}

print(my_set)
print(type(my_set))  # Output: <class 'set'>
```

```python
my_empty_set = set([1, 2, 3, 4]) # Create an empty set using the set() constructor
print(my_empty_set)
```

```python
my_set = set([1, 2, 3, 4, 5, 4, 6, 7, 5, 2]) # Create a set from a list with duplicate elements
print(my_set) # prints only unique elements
```

Set Operations

1. Adding and Removing Elements

```python
my_set.add(8)
my_set.add(9)
my_set.add(9) # this will not add 9 again since it's already present in the set
my_set
```

```python
my_set.remove(1)
print(my_set)
```

```python
my_set.discard(11) # remove element only if present
my_set
```

pop method

```python
removed_element = my_set.pop() # remove and return an arbitrary element from the set
print(removed_element)
print(my_set)
```

Clear all the elements

```python
my_set.clear()
print(my_set)
```

Set Membership test

```python
sett = {1, 2, 3,4 , 5, 6, 7}

print(3 in sett) # Output: True
print(9 in sett) # Output: False
```

Mathematical operation

```python
set1 = {1, 2, 3, 4, 5, 6}
set2 = {4, 5, 6, 7, 8, 9, 10, 11}

# Union - combines all unique elements from both sets
union_set = set1 | set2
print("Union:", union_set)


# Union using .union method
union_set_method = set1.union(set2)
print("Union using method:", union_set_method)
```

```python
set1 = {1, 2, 3, 4, 5, 6, 9}
set2 = {4, 5, 6, 7, 8, 9, 10, 11}

# Intersection - finds common elements between both sets
intersection_set = set1 & set2
print("Intersection:", intersection_set)


# Intersection using .intersection method
intersection_set_method = set1.intersection(set2)
print("Intersection using method:", intersection_set_method)


# Intersection Update - updates set1 to keep only elements found in both sets
intersection_update_method = set1.intersection_update(set2)
print("Intersection Update:", set1)  # set1 is now updated to the intersection of set1 and set2

print("Set1 After intersection operation", set1)
print("Set2 After intersection operation", set2)
```

```python
set1 = {1, 2, 3, 4, 5, 6, 9}
set2 = {4, 5, 6, 7, 8, 9, 10, 11}

# Difference - finds elements in set1 that are not in set2
difference_set = set1 - set2
print("Difference:", difference_set)


# Difference using .difference method
difference_set_method = set1.difference(set2)
print("Difference using method:", difference_set_method)


# Symmetric difference - finds elements in either set1 or set2 but not in both
symmetric_difference = set1.symmetric_difference(set2) 
print("Symmetric Difference:", symmetric_difference)
```

#### Sets Methods

```python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

# is subset - checks if set1 is a subset of set2, return true/false
check_is_subset =set1.issubset(set2)
print("Is set1 a subset of set2?  --> ", check_is_subset)

# is superset - checks if set1 is a superset of set2, return true/false
check_is_superset = set1.issuperset(set2)
print("Is set1 a superset of set2? --> ", check_is_superset)
```

```python
###  Counting Unique words in text

text = "This is a sample text with some sample words. This text is for testing purposes."

words = text.split()  # Split the text into words
print("Words:", words)

unique_words = set(words)  # Convert the list of words to a set to get unique words
print("Unique words:", unique_words)

print("Number of unique words:", len(unique_words))
```
