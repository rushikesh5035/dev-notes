- Dictionaries are unordered collections of items
- they store data in key-value pairs, keys must be unique and immutable(e.g., strings, numbers or tuples), while values can be of any

1. Creating Dictionaries

```python
empty_dict = {}

print(empty_dict)  # Output: {}
print(type(empty_dict))  # Output: <class 'dict'>
```

```python
# using the dict() constructor
empty_dict = dict()
empty_dict
```

```python
students={
    "name": "John",
    "age": 25,
    "courses": ["Math", "CompSci"]
}

print(students)  # Output: {'name': 'John', 'age': 25, 'courses': ['Math', 'CompSci']}
print(type(students))
```

```python
## Error -> duplicate key will not consider and it will consider the latest key's value
students={
    "name": "John",
    "age": 25,
    "name": 23
}

print(students)
```

2. Accessing Dictionary Elements

```python
students={
    "name": "John",
    "age": 25,
    "grade": "A"
}
```

```python
print("Studnets grades are:", students['grade'])
print("Students age is:", students['age'])
```

```python
## Accessing using get() method
print(students.get("grade"))

print(students.get("last_name")) # return none, key not present

print(students.get("last_name", "Not Available")) # key not present, default value
```

3. Modifying Dictionary Elements
- Dictionary are mutable, so we can add, update or delete elements

```python
print(students)
```

```python
students['age'] = 25
print(students)

students['address'] ="Pune"
print(students)
```

- delete the key and value pair

```python
del students["address"] # using del 
print(students)
```

4. Dictionary Methods

```python
# 1. .keys() - get all the keys
keys = students.keys()
print(keys)

# 2. .values() - get all the values
values = students.values()
print(values)

# 3. .items() - get all the items
items = students.items()
print(items)
```

- Shallow Copy

```python
students_copy = students
print(students)
print(students_copy)
```

```python
students["name"] = "Mike"
print(students)
print(students_copy)

# it updateds students name in both dictionaries because both are pointing to same memory location
```

```python
students["name"] = "Rushi"
print(students)
print(students_copy)
```

```python
#shallow copy -> create separate copy of the dictionary and will not point to the same memory location
students_copy1 = students.copy() # this will create a new copy of the dictionary and will not point to the same memory location

print(students_copy1)
print(students)
```

```python
students["name"] = "Chikku"
print(students_copy)
print(students)
```

5. Iterating Over Dictionaries

```python
## Iterating over keys

for keys in students.keys():
    print(keys)
```

```python
## Iterating over values

for values in students.values():
    print(values)
```

```python
## Iterating over items(key-value pair)

for key, value in students.items():
    print(f"{key}:{value}")
```

6. Nested Dictionaries 

```python
students = {
    "student1": {
        "name": "Rushi",
        "age": 23,
        "city": "Pune"
    },
    "student2": {
        "name":"Omkar",
        "age":22,
        "City":"Mumbai"
    }
}

students
```

- Accessing Nested dictionaries elements

```python
print(students["student1"]["name"])  # Output: Rushi
print(students["student1"]["age"])  # Output: Rushi

print(students["student2"]["name"])  # Output: omkar
```

- Iterating over nested dictionary

```python
for student_id, student_info in students.items():
    print(f"Student: {student_id}:{student_info}")
    for key, value in student_info.items():
        print(f"{key}:{value}")
        
    print()
```

7. Dictionary Comphrehension

```python
squares = { x:x**2 for x in range(5)} # x:x -> first x will be key and 2nd x will be value

squares
```

- Condition dictionary comphrehension

```python
even = {x:x**2 for x in range(10) if x%2==0}
even
```

### Practical Examples

- use a dictionary to count the frequency of elements in list

```python
numbers = [1, 2, 2, 2, 3, 4, 4, 4, 5, 6, 6, 6]

freq = {}

for num in numbers:
    if num in freq:
        freq[num] += 1
    else:
        freq[num] = 1

freq
```

- Merger 2 dictionaries into one

```python
dict1 = {"a": 1, "b": 2, "c": 3}
dict2 = {"b": 4, "c": 5, "d": 6}

merged_dict = {**dict1, **dict2}

print(f"Merged Dictionary: {merged_dict}")  # Output: {'a': 1, 'b': 4, 'c': 5, 'd': 6}
```
