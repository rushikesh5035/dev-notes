Polymorphism

- polymorphism is a core concept in OOP that allows objects of different classes to be treated as objects of a common superclass.
- ut provides a way to perform a single action in different forms.
- polymorphism is typically achieved through method **overriding** and **interfaces**

#### Method Overriding

- Method overring allows a child class to provide a specific implementation of a method that is already defined in its parent class. The implementation in the child class overrides the implementation in the parent class.

```python
## Method Overriding

# Base class

class Animal:
    def speak(self):
        return "Animal speaks"

# derived class 1
class Dog(Animal):
    def speak(self):
        return "Dog barks"

# derived class
class Cat(Animal):
    def speak(self):
        return "Meow!"

# Function that demonstrate polymorphism
def animal_speak(animal):
    print(animal.speak())
```

```python
dog = Dog()
cat = Cat()

print(dog.speak())
print(cat.speak())

animal_speak(dog)
animal_speak(cat)
```

```python
## Polymorphism with Functions and Methods

# Base class
class Shape:
    def area(self):
        return "The area of the figure"

# Derived class 1
class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height
    
    # overriding the area method
    def area(self):
        return self.width * self.height

# Derived class 2
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    
    # overriding the area method for Circle
    def area(self):
        return 3.14 * self.radius * self.radius


## Functions that demonstrate polymorphism

def print_area(shape):
    print(f"the area is {shape.area()}")

## Create a object
rectangle = Rectangle(4, 5)
circle = Circle(3)

print_area(rectangle)
print_area(circle)
```

#### Polymorphism with Abstract Base classes (ABSc)

- abstract base classes (ABSc) are used to define common methods for a related objects. 
they can enforce that derived classes implement particulara methods, promoting consistency across different implementations.

```python
from abc import ABC, abstractmethod

# abc - Abstract Base Class - a class that cannot be instantiated and is used to define a common interface for its subclasses.
# abstractmethod - empty method that is declared in the abstract class and must be implemented in the derived class.

## Define Abstract class

class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass

# Derived class 1
class Car(Vehicle):
    def start_engine(self):
        return "Car engine started"

# Derived class 2
class Motorcycle(Vehicle):
    def start_engine(self):
        return "Motorcycle engine started"

# Function that demonstrate polymorphism
def start_vehicle(Vehicle):
    print(Vehicle.start_engine())

# create object of Car and Motorcycle
car = Car()
motorcycle = Motorcycle()

start_vehicle(car)
start_vehicle(motorcycle)
```
