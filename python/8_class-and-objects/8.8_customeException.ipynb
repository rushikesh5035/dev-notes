Custom Exception (Raise and Throw an exception)

```python
# Create custom exception class

class Error(Exception):
    pass


class dobException(Error):
    """Raised when the input date of birth is invalid"""
    pass
```

```python
year = int(input("Enter your year of birth: "))

age = 2026 - year
try:
    if age <= 30 and age >= 20:
        print("You are eligible to apply for this job.")
    else:
        raise dobException # here we are raising the exception if the age is not in the range of 20 to 30

except dobException: # and here we are catching the exception and printing the message
    print("Sorry, you are not eligible to apply for this job. Your age should be greater than 20 and less than 30.")
```
