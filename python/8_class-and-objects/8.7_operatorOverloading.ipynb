Operator Overloading

- Operator overloading allows you to define the behavior of operators (+, -, *, etc), for custom objects.
- achieve this by overriding specific magic method in your class

```python
### Common Operator Overriding Magic Methods

'''
__add__(self, other) - add two objects using the + operator
__sub__(self, other) - subtract two objects using the - operator
__mul__(self, other) - multiply two objects using the * operator
__truediv__(self, other) - divide two objects using the / operator
__eq__(self, other) - check if two objects are equal using the == operator
__lt__(self, other) - check if one object is less than another using the < operator
--gt__(self, other) - check if one object is greater than another using the > operator
'''
```

```python
## Mathematical operations for vectors using operator overloading
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # Implementing the __repr__ method to return a string representation of the vector (without this method, the output of the vector object will be <__main__.Vector object at 0x0000018DC2281D10>)
    def __repr__(self):
        return f"Vector({self.x}, {self.y})"

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y) # Add two vectors by adding their corresponding components

    def __sub__(self, other):
        return Vector(self.x - other.x, self.y - other.y) # Subtract two vectors by subtracting their corresponding components

    def __mul__(self, other):
        return Vector(self.x * other.x, self.y * other.y) # Multiply two vectors by multiplying their corresponding components

    def __truediv__(self, other):
        return Vector(self.x / other.x, self.y / other.y) # Divide two vectors by dividing their corresponding components

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y # Check if two vectors are equal by comparing their corresponding components 

    def __lt__(self, other):
        return self.x < other.x and self.y < other.y # Check if one vector is less than another by comparing their corresponding components

    def __gt__(self, other):
        return self.x > other.x and self.y > other.y # Check if one vector is greater than another by comparing their corresponding components


# create objects of the Vector class
v1 = Vector(2, 3)
v2 = Vector(4, 5)

# print(v1+v2) # Output: <__main__.Vector object at 0x0000018DC2281D10>
# so to fix the output we need to implement the __repr__ method in the Vector class. 
# The __repr__ method is a special method that is called when we try to print an object. It should return a string representation of the object.

print(v1 + v2) # Output: Vector(6, 8)

# v1 will be considered as self and v2 will be considered as other in the __add__ method. The method will return a new Vector object with x and y values equal to the sum of the corresponding x and y values of v1 and v2.
```

```python
print(v1 - v2) # Output: Vector(-2, -2)
print(v1 * v2) # Output: Vector(8, 15)
print(v1 / v2) # Output: Vector(0.5, 0.6)
print(v1 == v2) # Output: False
print(v1 < v2) # Output: True
print(v1 > v2) # Output: False
```

```python
### Overloading operators for complex numbers

class ComplexNumber:
    def __init__(self, real, imag):
        self.real = real
        self.imag = imag

    def __repr__(self):
        return f"ComplexNumber({self.real}, {self.imag}i)"

    def __add__(self, other):
        return ComplexNumber(self.real + other.real, self.imag + other.imag)

    def __sub__(self, other):
        return ComplexNumber(self.real - other.real, self.imag - other.imag)

    def __mul__(self, other):
        real_part = self.real * other.real - self.imag * other.imag
        imag_part = self.real * other.imag + self.imag * other.real
        return ComplexNumber(real_part, imag_part)

    def __truediv__(self, other):
        denom = other.real ** 2 + other.imag ** 2
        real_part = (self.real * other.real + self.imag * other.imag) / denom
        imag_part = (self.imag * other.real - self.real * other.imag) / denom
        return ComplexNumber(real_part, imag_part)

    def __eq__(self, other):
        return self.real == other.real and self.imag == other.imag

    def __lt__(self, other):
        return (self.real ** 2 + self.imag ** 2) < (other.real ** 2 + other.imag ** 2)

    def __gt__(self, other):
        return (self.real ** 2 + self.imag ** 2) > (other.real ** 2 + other.imag ** 2)

# create objects of the ComplexNumber class
c1 = ComplexNumber(2, 3)
c2 = ComplexNumber(4, 5)

# use overloaded operators

print(c1 + c2) # Output: 6 + 8i
print(c1 - c2) # Output: -2 - 2i
print(c1 * c2) # Output: -7 + 22i
print(c1 / c2) # Output: 0.44 + 0.08i
print(c1 == c2) # Output: False
print(c1 < c2) # Output: True
print(c1 > c2) # Output: False
```
