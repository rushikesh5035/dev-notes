Magic Methods

Magic methods in pythin also known as dunder methods(double underscore methods), are special methods that start and end with double underscore.
These methods enable you to define the behavior of objects for built-in perations, such as arithmetic operations, comparisons and more

```python
"""
Magic Methods

- Magic methods are predefined methods in python that you can override to change the behavior of your objects, Some common magic methods includes:

1.  __init__: initializes a new instance of a class
2.  __str__: Returns a string representation of an object
3.  __repr__: Returns a offial string representation of an object
4.  __len__: Return the length of an object
5.  __getitem__: gets an item from a container.
6.  __setitem__: sets an item in a container.
"""
```

```python
class Person:
    pass

person = Person()

dir(person) # it shows all the magic mathods
```

```python
print(person) # output print from __str__
```

```python
## Basics methods

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

person = Person("Rushi", 23)

print(person)
# <__main__.Person object at 0x0000024F4D8FFE00>

# we can override this output using __str__
```

```python

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"{self.name}, {self.age} years old"

    # custom
    def __repr__(self):
        return f"Person (name={self.name}, age={self.age})"

person = Person("Rushi", 23)

print(person)
# output --> Rushi, 23 years old

print(repr(person))
```
