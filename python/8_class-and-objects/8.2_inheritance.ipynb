Inheritance in Python

- Inheritance is a fundamental concept in OOP that allows a class to **inherit attributes and methods from another class**
- Single Inheritance
- Multiple Inheritance

Single Inheritance

```python
## Inheritance

# Parent Class
class Car:
    def __init__(self, windows, doors, enginetype):
        self.windows = windows
        self.doors = doors
        self.enginetype = enginetype
    
    def drive(self):
        print(f"The person will drive the {self.enginetype} car")
```

```python
car1 = Car(4, 5, "petrol")
car1.drive()
```

```python
# suppose we have this Tesla car class, it has some properties like windows, doors, engintype, color, currentSpeed etc, so we need to define this properties, but if you see carefully in the car class we have already defined some of the properties like windows, doors, enginetype, so instead of defining them again in the Tesla class we can inherit the Car class and use its properties in the Tesla class.

# make tesla inherit class
class Tesla(Car): 
    def __init__(self, windows, doors, enginetype, color, currentSpeed, is_selfdriving):
        super().__init__(windows, doors, enginetype) # super() is used to call the parent class constructor(so we will use parent class properties here)
        self.color = color
        self.currentSpeed = currentSpeed
        self.is_selfdriving = is_selfdriving

    def selfDriving(self):
        print(f"Tesla supports self driving - {self.is_selfdriving}")
    
    def carDetails(self):
        print(f"Tesla car's features -> "
              f"Windows: {self.windows}, "
              f"Doors: {self.doors}, "
              f"Engine Type: {self.enginetype}, "
              f"Color: {self.color}, "
              f"Current Speed: {self.currentSpeed}, "
              f"Is Self Driving: {self.is_selfdriving}")
```

```python
tesla1 = Tesla(4, 5, "eletric", "white", 120, False)
tesla1.selfDriving()
```

```python
tesla1.drive()
```

```python

tesla1.carDetails()
```

Multiple Inheritance
- when a class inherits from more than one base class

```python
# Base Class
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print("Subclass must implement this method")

# Base Class 2
class Pet:
    def __init__(self, owner):
        self.owner = owner

# Derived Class
class Dog(Animal, Pet):
    def __init__(self, name, owner):
        Animal.__init__(self, name) # inheriting the Animal class attribute
        Pet.__init__(self, owner)

    def speak(self):
        return f"{self.name} says Woof! and is owned by {self.owner}"
```

```python
# create an object
dog1 = Dog("Buddy", "Alice")

dog1.speak()
```
