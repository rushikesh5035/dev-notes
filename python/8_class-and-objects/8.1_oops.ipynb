Classes and Objects

- object oriented Programming (OOP) is programming paradigm that uses **Objects** to design applications and computer programs.
- OOP Allows for modeling real-world scenarios using classes and objects. 

```python
## A class is a blue print for creating objects. Attributes are variables that belong to a class. Methods are functions that belong to a class.

# Creating a class named Car
class Car:
    pass

audi = Car()  # Creating an object of the Car class
bmw = Car()   # Creating another object of the Car class
mercedes = Car()  # Creating yet another object of the Car class

print(audi) # Output: <__main__.Car object at 0x7f8b8c2e1d30> 
# 0x7f8b8c2e1d30 is the memory address where the object is stored. It will be different each time you run the code.

print(bmw) 
print(mercedes)

print(type(audi))  # Output: <class '__main__.Car'>
```

```python
audi.window = 4  # Adding an instance variable 'window' to the audi object
print(audi.window)  # Output: 4

# But initializing attributes like this is not a good practice.
```

```python
tata = Car()  # Creating an object of the Car class

tata.door = 4 
print(tata.door)  # Output: 4

dir(tata)  # Output: ['__class__', '__delattr__', '__dict__', '__dir__', '__doc__', '__eq__', '__format__', '__ge__', '__getattribute__', '__gt__', '__hash__', '__init__', '__init_subclass__', '__le__', '__lt__', '__module__', '__ne__', '__new__', '__reduce__', '__reduce_ex__', '__repr__', '__setattr__', '__sizeof__', '__str__', '__subclasshook__', '__weakref__', 'door']
```

```python
## Instance Variables

# Instance variables are variables that are defined inside a constructor and belong to the object. They are unique to each object.

class Dog: 

    ## Constructor - (self is a reference to the current instance of the class)
    def __init__(self, name, age):
        self.name = name
        self.age = age


# instance variables in above class are --> self.name and self.age

# Create object
dog1 = Dog("Buddy", 3)  # Creating an object of the Dog class

print(dog1) # Output: <__main__.Dog object at 0x7f8b8c2e1d30>

print(dog1.name)
print(dog1.age)
```

```python
dog2 = Dog("Max", 5)
dog3 = Dog("Lucy", 10)
```

```python
# Define Class with Instance methods 

# Instance methods  are functions that are defined inside a class and belong to the object. They can access and modify the instance variables of the object.

class Dog:

    # Constructor
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # Instance methods
    def bark(self):
        print(f"{self.name} is barking.")
    
    def get_age(self):
        return self.age
    

dog1 = Dog("Buddy", 3)

dog1.bark()

print(dog1.get_age())  # Output: 3

dog2 = Dog("Max", 5)

dog2.bark()
print(dog2.get_age())
```

#### Modeling a Bank Account

```python
# Define a class for bank account

class BankAccount:

    # Constructor
    def __init__(self, account_holder, account_number, balance=0):
        self.account_holder = account_holder
        self.account_number = account_number
        self.balance = balance

    # instance methods
    def deposit(self, amount):
        if amount <= 0:
            print("Deposit amount must be positive.")
            return

        self.balance += amount
        print(f"{amount} is deposited. New balance is {self.balance}")
    
    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient balance.")
        else:
            self.balance -= amount
            print(f"{amount} is withdrawn. New balance is {self.balance}")

    def get_balance(self):
        return self.balance

## Create an account
account = BankAccount("Rushikesh", 12345, 5000)

account.get_balance()  # Output: 5000
```

```python
account.deposit(2000)  # Output: 2000 is deposited. New balance is 7000
```

```python
account.withdraw(1000)  # Output: 1000 is withdrawn. New balance is 6000
```
